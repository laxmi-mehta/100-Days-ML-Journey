```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('../../Datasets/train (1).csv')
```

```python
df.head()
```

```python
X = df.iloc[:,1:]
y = df.iloc[:,0]
```

```python
sns.heatmap(X.select_dtypes(include='number').iloc[5].values.reshape(1, -1), annot=True)

X = X.select_dtypes(include='number')
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()

rf.fit(X,y)
```

```python
rf.feature_importances_
```

```python
rf.feature_importances_.shape
```

```python
n = rf.feature_importances_.shape[0]

sns.heatmap(rf.feature_importances_.reshape(1,n))
```
